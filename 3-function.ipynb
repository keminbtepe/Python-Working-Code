```python
def sozluk(**kwargs):   #directory ataması yapabilriz istedigimiz kadar
    if "apple" in kwargs:
        p("apple var içeride")
    else:
        p("apple yok icerisinde")
```

```python
myString = ["kursat","besıktepe","emin","kursat besiktepe","kursat emin"]
```

```python
def controlString(string):
    return "kursat" in string
```

```python
controlString("kursat")
```

## MAP 

```python
list(map(controlString,myString))
```

## FİLTER

```python
list(filter(controlString,myString))  #icinde kursat geçenleri getirir
```
