## Merge 

```python
df_merge1 = pd.read_csv("7-merge_data1.csv")
```

```python
df_merge1
```

```python
df_merge2 = pd.read_csv("7-merge_data2.csv")
```

```python
df_merge2
```

```python
## merge - outer join (iki tabloyu ıdye göre birleştirme) iki tabloyuda alır
```

```python
df_merge_outer = pd.merge(df_merge1,df_merge2, on="Employee_ID",how="outer")  # on = ortak olan ıdye göre eşler
```

```python
df_merge_outer
```

```python
#merge - left-right join 
```

```python
df_merge_left = pd.merge(df_merge1,df_merge2,on="Employee_ID",how="left")  # on = ortak olan ıdye göre eşler
```

```python
df_merge_left # 1.tablodaki bütün elemanları alır 2.tablodaki boş değerleri(eşleşmeyen) almaz
```

```python
dp_merge_right = pd.merge(df_merge1,df_merge2,on="Employee_ID",how="right")
```

```python
dp_merge_right
```

```python
# merge - inner join (her iki tablonunda kesişim kümesini alır boş satır almaz)
```

```python
df_merged = pd.merge(df_merge1,df_merge2,on="Employee_ID",how="inner")
```
