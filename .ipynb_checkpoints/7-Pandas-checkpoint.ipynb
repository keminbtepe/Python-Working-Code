```python
import numpy as np
import pandas as pd
```

## SERİES 

```python
benimSozlugum = {"Kürsat":50, "Zeynep":40, "Mehmet":30}
```

```python
pd.Series(benimSozlugum)
```

```python
benimYaslarim = [50,40,30]
benimIsimlerim = ["Kürsat", "Zeynep","Mehmet"]
```

```python
pd.Series(data=benimYaslarim,index=benimIsimlerim) #data index kısmını yazmadan degiskenleri yazsakda olur
```

```python
numpyDizisi = np.array([50,40,30])
```

```python
pd.Series(numpyDizisi,benimIsimlerim)
```
