```python
class Meyve():
    def __init__(self,isimInput,kaloriInput):
        self.isim = isimInput
        self.kalori = kaloriInput
    def __str__(self): ##str yazdırma komutu gibi bir şey print dedigimzde çalısan
        return f"{self.isim} kalorisi: {self.kalori}"
    def __len__(self):
        return self.kalori
```

```python
muz = Meyve("Muz",150)
```

```python
muz.kalori
```

```python
print(muz)
```

```python
len(muz)
```
