```python
import pandas as pd
import numpy as np
```

```python
sozlükVeri = {"Ist":[30,29,np.nan],"Ank":[20,np.nan,25],"Izm":[40,39,38]}
havaDurumuDataFrame = pd.DataFrame(sozlükVeri,index=["Ptesi","Salı","Cars"])
```

```python
havaDurumuDataFrame
```

```python
havaDurumuDataFrame.dropna() #boş olan satırları otomatik siler row axix=0
```

```python
havaDurumuDataFrame.dropna(axis=1) #column
```

```python
sozlükVeri1 = {"Ist":[30,29,np.nan],"Ank":[20,np.nan,25],"Izm":[40,39,38],"Ant":[45,np.nan,np.nan]}
yeniDataFrame = pd.DataFrame(sozlükVeri1,index=["Ptesi","Salı","Cars"])
```

```python
yeniDataFrame
```

```python
yeniDataFrame.dropna(axis=1,thresh=2) #2 ve üzeri nan olanları atar 
```

```python
yeniDataFrame.loc["Salı"].fillna(20)  # boş olanlara 20 degerini yazar
```

```python
yeniDataFrame
```

```python
yeniDataFrame["Ist"] = yeniDataFrame["Ist"].fillna(35.5)
yeniDataFrame   ## belli sütundaki boşlukları doldurma kodu
```

```python
yeniDataFrame.loc["Salı"] = yeniDataFrame.loc["Salı"].fillna(45)   #salıdakileri 45 yapar
```

```python
yeniDataFrame
```

## GROUPBY 

```python
maasSozlugu = {"Departman" : ["Yazılım","Yazılım","Pazarlama","Pazarlama","Hukuk","Hukuk","Hukuk","Sağlık"],
              "Calisan İsmi" : ["Ali","Mehmet","Turan","Samet","Zeynep","Kader","Yavuz","Ümit"],
               "Maas" : [100,200,300,400,500,600,700,800]
              }
```

```python
maasDf = pd.DataFrame(maasSozlugu)
```

```python
maasDf
```

```python
grup1 = maasDf.groupby("Departman")
```

```python
grup1.count()
```

```python
grup1.describe()
```

```python
grup1.mean(numeric_only=True) #sayısal sütunları seçiyoruz öbür türlü hata veriyor
```

```python
grup1.count().mean()
```
