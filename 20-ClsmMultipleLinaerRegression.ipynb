```python
#X = df.iloc[:,:-1]
#y = df.iloc[:,-1]

X = df[["Study Hours","Sleep Hours","Attendance Rate","Social Media Hours"]]
y = df["Exam Score"]
```

```python
#train Test Split
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25, random_state=15)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
#standartlaştırma
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
regression = LinearRegression()
```

```python
regression.fit(X_train, y_train)
```

```python
X_test[0]
```

```python
df.iloc[0]
```

```python
# tahmin etmek için degişkenleri girdigimiz bir öğrenci oluşturduk
new_student = [[5,7,90,2]]   # 5 saat calıstı 7 saat uyudu 
# derslerin %90ına katıldı sosyal medyada 2 saat geçirdi
```

```python
new_student_scaled = scaler.transform(new_student) #standartlaştırdık
```

```python
regression.predict(new_student_scaled)  # bu arkadas 90 puan alır olarak tahmın etti
```

## Prediction Başarısını ölçüyoruz

```python
y_pred = regression.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error, r2_score
```

```python
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
print("MSE: ", mse)
print("MAE: ", mae)
```

```python
score = r2_score(y_test,y_pred)
print("r2 score:", score)
```

```python
#adjusted r2 score (düzeltilmiş r kare)
1 - (1-score)* (len(y_test)-1) / (len(y_test)-X_test.shape[1]-1)
```
