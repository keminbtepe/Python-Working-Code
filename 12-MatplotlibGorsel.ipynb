```python
age_list = [3,6,7,8,8,12,15,16,17,18,19,21,22,23,24,25]
```

```python
weight_list = [10,20,30,30,42,43,54,56,68,72,74,77,82,85,88,77]
```

```python
print(f"weight:{len(weight_list)} age:{len(age_list)}")
len(weight_list) == len(age_list)  #eşit olması lazım
```

```python
plt.plot(age_list,weight_list,"r") # y -> yellow
plt.xlabel("Age")
plt.ylabel("Weight")
plt.title("Age vs Weight")
#plt.show()  matplotlib inline yazdıgımız için otomatik olarak gösterir
#plt.savefig("kaydetti.png")
```

```python
numpy_arr1 = np.round(np.linspace(0,10,9))
numpy_arr1.astype(int)
```

```python
numpy_arr2 = numpy_arr1 ** 3
```

```python
plt.plot(numpy_arr1,numpy_arr2,"b.--")
plt.show()
```

```python
plt.plot(numpy_arr2,numpy_arr1,"r.--")
```

```python
plt.subplot(1,2,1)  #1 satır row 2 column sütun 1. grafik
plt.plot(numpy_arr1,numpy_arr2,"b.--")
plt.xlabel("xlabel")
plt.ylabel("ylabel")
plt.title("baslık")
plt.subplot(1,2,2)
plt.plot(numpy_arr2,numpy_arr1,"r.--")
plt.xlabel("xlabel")
plt.ylabel("ylabel")
plt.title("baslık2")
```

```python
mfigure = plt.figure()
figure_axes = mfigure.add_axes([0.1,0.1,0.4,0.4])  #0.4'ler en boy oranı 
figure_axes.plot(numpy_arr1,numpy_arr2,"r.--")
figure_axes.set_xlabel("X ekseni")
figure_axes.set_ylabel("Y ekseni")
figure_axes.set_title("Baslık")
plt.show()
```

```python
new_fig = plt.figure(dpi=75)
new_axes = new_fig.add_axes([0.1,0.1,0.9,0.9])
new_axes.plot(numpy_arr1,numpy_arr1**2,label = "np array **2")
new_axes.plot(numpy_arr1,numpy_arr1**3,label = "np array **3")
new_axes.legend(loc=6)
```
