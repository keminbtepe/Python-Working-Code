## İNHERİTANCE

```python
class Musician():

    def __init__(self,name):
        self.name = name
        print("musician class")
    def test1(self):
        print("sarki söyleybilir")
    def test2(self):
        print("araba süreiblir")
        
```

```python
emin = Musician("emin")
```

```python
emin.name
```

```python
emin.test1()
```

```python
class MusicianPlus(Musician):   ## musicianplus içine musician özelliklerini verdik besiktepe hep musician hemde musician plus özellikli

    def __init__(self,name):
        Musician.__init__(self,name)
        print("musician plus")
    def ozellik1(self):
        print("ameliyat yapabilir")
```

```python
besiktepe = MusicianPlus("besiktepe")
```

```python
besiktepe.test1()
```

```python
besiktepe.ozellik1()
```

```python
class Banana():
    def __init__(self,name):
        self.name = name
    def info(self):
        print(f"100 kalori: {self.name}")
class Apple():
    def __init__(self,name):
        self.name = name
    def info(self):
        print(f"150 kalori: {self.name}")
```

```python
banana = Banana("banana")
```

```python
banana.info()
```

```python
apple = Apple("apple")
```

```python
apple.info()
```

```python
fruitList = [banana,apple]
```

```python
for fruit in fruitList:
    print(fruit.info())
```

## ENCAPSULATION

```python
class Phone():
    def __init__(self,name,price):
        self.name = name
        self.__price = price
    def info(self):
        print(f"{self.name} ürününün fiyatı {self.__price}TL dir")
    def changePrice(self,price):
        self.__price = price
```

```python
iphone = Phone("İphone 14 PRO",900)
```

```python
iphone.info()
```

```python
iphone.price = 850   # __price yaptıgımızdan degistirilemez kılar
```

```python
iphone.info()
```

```python
iphone.changePrice(850)  ## yanlıslıkla degistirilmesin diye ayrı fonksiyonda yetki verdik
```

```python
iphone.info()
```

## ABSTRACTION

```python
from abc import ABC, abstractmethod
```

```python
class Car(ABC):
    
    @abstractmethod
    def MaxSpeed(self):
        pass
```

```python
myCar = Car()  #oluşturmaz çünkü sınıf açıp içinde Car sınıfının özelliklerini kullanabiliriz sadece
```

```python
class Tesla(Car):

    def MaxSpeed(self):
        print("150 km/h")
```

```python
tesla = Tesla()
```

```python
tesla.MaxSpeed()
```

```python
class Mercedes(Car):

    def MaxSpeed(self):
        print("250 km/h")
    def __str__(self):   ##print olarak kullandıgımda string buna döner
        return "selam"
```

```python
mercedes = Mercedes()
```

```python
mercedes.MaxSpeed()
```

```python
print(Mercedes())
```
