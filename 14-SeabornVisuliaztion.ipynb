```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
data = pd.read_csv("athlete_events.csv")
```

```python
data.head()
```

```python
data.describe()
```

```python
data.info()
```

```python
data.count()
```

```python
data.columns
```

```python
plt.scatter("Height","Weight",data=data)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Boy Kilo Karşılaştırması")
plt.show()
```

```python
sns.set
```

```python
sns.set_style("darkgrid") #stili 
sns.scatterplot(x="Height",y="Weight",hue="Sex",style="Medal",data=data)  #hue cinsiyetleri ikiye ayırır renk renk
plt.xlabel("Height Of Atheletes")
plt.ylabel("Weight Of Athelets")
plt.title("Athletes Weight Vs Height")   #snsnin farkı noktaların çevresinde beyaz çizgi var
```

```python
data.columns
```

```python
data["Medal"].unique()  #madalyaların içerdeigi seçenekleri gösterir
```

```python
data["Sex"].unique()
```

```python
sns.set_style("darkgrid") #stili 
sns.scatterplot(x="Height",y="Weight",hue="Sex",size="Age",data=data)  #hue cinsiyetleri ikiye ayırır renk renk
plt.xlabel("Height Of Atheletes")
plt.ylabel("Weight Of Athelets")
plt.title("Athletes Weight Vs Height")   #snsnin farkı noktaların çevresinde beyaz çizgi var
```

```python
sns.set_style("whitegrid")
sns.lineplot(x="Height",y="Weight",hue="Sex",data=data) #bu grafigin kullanımı
plt.xlabel("Height Of Atheletes")
plt.ylabel("Weight Of Athelets")
plt.title("Athletes Weight Vs Height")
plt.show()
```

```python
sns.set_style("white")
sns.displot(x="Height",data=data)
plt.ylabel("Frequency")
plt.show()
```

```python
sns.set_style("darkgrid")
sns.kdeplot(x="Height",hue="Sex",multiple="stack",data=data)
plt.ylabel("Weight")
plt.show()
```

```python
teamsGroup = data.groupby("Team")
```

```python
sns.barplot(x="Medal",y="Height",hue="Sex",data=data)
plt.show()
```
