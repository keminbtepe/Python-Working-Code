```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data1 = np.linspace(0,10,20)
```

```python
data1
```

```python
data2 = data1 **2 
```

```python
data2
```

```python
plt.subplots()
```

```python
my_fig, my_axes = plt.subplots()
```

```python
my_fig, my_axes = plt.subplots()
my_axes.plot(data1,data2,alpha = 0.5) #alpha transparan oranı
my_axes.plot(data2,data1)
plt.show()
```

```python
(new_fig, new_axes) = plt.subplots()
new_axes.plot(data1,data1+2 ,linewidth = 3,color="Red")
new_axes.plot(data1,data1+4 ,linewidth =4,color="Blue",linestyle="-.")
new_axes.plot(data1,data1+5 ,linewidth =4,color="Green",linestyle="--")
new_axes.plot(data1,data1+6 ,linewidth =4,color="yellow",linestyle=":")
new_axes.plot(data1,data1+7 ,linewidth =4,color="Black",linestyle=":",marker="o",markersize=8,markerfacecolor="red")
new_axes.plot(data1,data1+8 ,linewidth =4,color="Purple",linestyle=":",marker="+",markersize=10)
```
