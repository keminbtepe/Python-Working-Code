```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder
```

```python
df = sns.load_dataset("titanic")
```

```python
df  #survived sütunu kurtulamadı kurtuldu
```

```python
df[["sex","class","embark_town"]].isnull().sum()
```

```python
df = df.dropna(subset="embark_town")
```

```python
df[["sex","class","embark_town"]].isnull().sum()
```

## one hot encoding

```python
df["sex"].value_counts()
```

```python
df["embark_town"].value_counts()
```

```python
df["class"].value_counts()
```

```python
df_onehot = pd.get_dummies(df,columns=["sex","embark_town"],drop_first=True)
```

```python
df_onehot  #sex_male diye sütun olusturup true false degerlerini
#verdi queenstown da yasamıyorsa false suthampton da yasamıoyr ise false
#sonuc olarak Cherbourgda yasamıs oluyor
```

```python
df_onehot.columns
```

## label encoder

```python
label_encoder = LabelEncoder()
```

```python
df_label = df.copy()
```

```python
df_label
```

```python
label_encoder.fit_transform(df_label["sex"]) #1-0 yapıyor
```

```python
df_label["sex"] = label_encoder.fit_transform(df_label["sex"])
```

```python
df_label.head() #1 - 0 lı oldu
```

## ordinal encoder

```python
df_ordinal = df.copy()
```

```python
class_order = ["Third","Second","First"]
```

```python
ordinal_encoder = OrdinalEncoder(categories=[class_order])
```

```python
df_ordinal["class"] = ordinal_encoder.fit_transform(df_ordinal[["class"]])
```

```python
df_ordinal.head()
```

```python
fig, axes = plt.subplots(1,3,figsize=(15,3))  #hepsini karsılastırıyoruz
df["sex"].value_counts().plot(kind="bar",ax=axes[0],title="original categorical")
df_label["sex"].value_counts().plot(kind="bar",ax=axes[1],title="original categorical")
df_onehot["sex_male"].value_counts().plot(kind="bar",ax=axes[2],title="original categorical")
```
