## inheritance

```python
class Hayvan():
    def __init__(self):
        print("hayvan sınıfı init çagırıldı")
    def method1(self):
        print("hayvan sınıfı method1")
    def method2(self):
        print("hayvan sınıfı method2")
```

```python
benimHayvanım = Hayvan()
```

```python
benimHayvanım.method2()
```

```python
class Kedi(Hayvan):
    def __init__(self):
        Hayvan.__init__(self)
        print("kedi sınıfı init çagırıldı")
    def miyavla(self):
        print("miyavla")
    ##OVERRİDE (ÜZERİNE YAZMAK)
    def method1(self):
        print("kedi sınıfı method1 çagırıldı")
```

```python
benimKedi = Kedi()
```

```python
benimKedi.miyavla()
```

```python
benimHayvanım.method1()
```

```python
digerKedi = Kedi()
```

```python
digerKedi.method1()
```

## Polymorphism

```python
class Elma():
    def __init__(self,isimInput):
        self.isim = isimInput
    def bilgiVer(self):
        return self.isim + " 100 kaloridir."
```

```python
class Muz():
    def __init__(self,isimInput):
        self.isim = isimInput
    def bilgiVer(self):
        return self.isim + " 150 kaloridir."
```

```python
elma = Elma("elma")
```

```python
elma.bilgiVer()
```

```python
muz = Muz("Muz")
```

```python
muz.bilgiVer()
```

```python
meyveListesi = [elma,muz]
```

```python
for meyve in meyveListesi:
    print(meyve.bilgiVer())
```
