```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
numpyDizisi1 = np.linspace(0,10,20)
```

```python
numpyDizisi1
```

```python
numpyDizisi2 = numpyDizisi1 **2 
```

```python
(benimFigure, benimEksen) = plt.subplots(dpi=100)
benimEksen.plot(numpyDizisi1,numpyDizisi2,color="#C96F23",alpha=0.3) #alpha çizginin şeffaflık degeridir
benimEksen.plot(numpyDizisi2,numpyDizisi1,"r")
```

```python
(yeniFigur, yeniEksen) = plt.subplots()
yeniEksen.plot(numpyDizisi1, numpyDizisi1+2,color="green",linewidth=8.0)  #linewidth çizgi kalınlıgıdır
yeniEksen.plot(numpyDizisi1, numpyDizisi1+3,"b")
```

## Scatter

```python
plt.scatter(numpyDizisi1,numpyDizisi2)
```
