## Döngüler 

```python
my_list = [10,20,30,40,50,60,70]

for x in my_list:
    if x % 6 == 0:
        print(x)
        
```

```python
for a in my_list:
    if a == 40:
        continue  # bu satırı atla diger sayıdan devam et
    print(a)
```

```python
x = 0
while x <= 3:  # x 10dan küçükse çalıştır
    print(x)
    x = x+1
print("döngü bitti")
```
