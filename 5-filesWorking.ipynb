```python
def Fprint(fileName):   #dosyayı okuma fonskiyonu read
    with open(fileName,mode="r") as myFile5:
        myContent2 = myFile5.read()
    return myContent2
```

```python
Fprint("myfile.txt")
```

```python
def Fappend(fileName,text):
    with open(fileName,mode="a") as myFile:
        myFile.write(text)
```

```python
workingFile = "myfile.txt"
Fappend(workingFile," /n function new text append")
Fprint(workingFile)
```

```python
def proces_value(value: int | str ) -> str:   #ya int ya string alıcak geriye string döndürecek demek
    if isinstance(value,int):   #value typesi integer mi
        return f"deger integerdir {value}"
    else:
        return f"deger string {value}"
```
