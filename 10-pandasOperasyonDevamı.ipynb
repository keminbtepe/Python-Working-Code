```python
import numpy as np
import pandas as pd
import os
```

```python
sozluk1 = {"Isim" : ["Ahmet","Mehmet","Zeynep","Kursat"],
              "Spor" : ["Koşu","Yüzme","Koşu","Basketbol"],
               "Kalori" : [100,200,300,400]
              }
```

```python
sozluk2 = {"Isim" : ["Osman","Levent","Kursat","Fatma"],
              "Spor" : ["Koşu","Yüzme","Koşu","Basketbol"],
               "Kalori" : [100,200,250,150]
              }
```

## MERGE (Kaynaştırma - Birleştirme gibi)

```python
mergeSozluk1 = {"Isim" : ["Ayse","Mahmut","Duygu","Nur"],
              "Spor" : ["Koşu","Yüzme","Koşu","Basketbol"]
              }
```

```python
mergeDataFrame1 = pd.DataFrame(mergeSozluk1)
```

```python
mergeSozluk2 = {"Isim" : ["Ayse","Mahmut","Duygu","Nur"],
              "Kalori" : [400,500,100,200]
              }
```

```python
mergeDataFrame2 = pd.DataFrame(mergeSozluk2)
```

```python
mergeDataFrame2
```

```python
mergeDataFrame1
```

```python
pd.merge(mergeDataFrame1,mergeDataFrame2,on="Isim")  # ortak kolon üzerinden birlestirme yapar
```

```python
maasSozlugu = {"Departman" : ["Yazılım","Yazılım","Yazılım","Pazarlama","Pazarlama"],
              "Calisan İsmi" : ["Ali","Ali","Mehmet","Turan","Samet"],
               "Maas" : [100,300,200,300,400]
              }
```

```python
maasDataFrame1 = pd.DataFrame(maasSozlugu)
```

```python
maasDataFrame1
```

```python
maasDataFrame1["Departman"].unique() #benzersiz kaç tane departman var 
```

```python
maasDataFrame1["Departman"].nunique()  # benzersiz sayısını yazar
```

```python
maasDataFrame1["Departman"].value_counts()  # her departmandan kaç adet oldugunu yazar
```

```python
def brutMaas(maas):
    return maas * 0.66
```

```python
maasDataFrame1["Maas"].apply(brutMaas)
```

```python
maasDataFrame1
```

```python
maasDataFrame1.pivot_table(values="Maas", index=["Departman","Calisan İsmi"],aggfunc=["sum","mean"])  
```
