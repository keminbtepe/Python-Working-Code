```python
def corelation_for_dropping(df, treshold):
    columns_to_drop = set()
    corr = df.corr()
    for i in range(len(corr.columns)):
        for j in range(i):
            if abs(corr.iloc[i,j]) > treshold:
                columns_to_drop.add(corr.columns[i])
    return columns_to_drop
```

```python
corelation_for_dropping(X_train, 0.85) #belli deger üzerindeki korelasyonu olan
#colonları bize getirir
```

```python
columns_dropping = corelation_for_dropping(X_train, 0.85)  #silecegimiz sütunları
#degiskene atıyoruz
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
X_train.drop(columns_dropping,axis=1,inplace=True)
X_test.drop(columns_dropping,axis=1,inplace=True)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
#datalarımızı temizledik şimdi regresyon aşamalarına başlayacagız
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
#normalizasyonu grafik üzerinde ele alıyoruz öncesi sonrası
plt.subplots(figsize =(15,5))
plt.subplot(1,2,1)
sns.boxplot(data=X_train)
plt.title("X_train")
plt.subplot(1,2,2)
sns.boxplot(data=X_train_scaled)
plt.title("X_train_scaled")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error, r2_score
```

```python
from sklearn.linear_model import Lasso
lasso = Lasso()
lasso.fit(X_train_scaled,y_train)
y_pred = lasso.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("R2 Score: ",score)
plt.scatter(y_test,y_pred)
plt.show()
```

```python
from sklearn.linear_model import Ridge
Ridge = Ridge()
Ridge.fit(X_train_scaled,y_train)
y_pred = Ridge.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("R2 Score: ",score)
plt.scatter(y_test,y_pred)
plt.show()
```

```python
from sklearn.linear_model import ElasticNet
Elasticnet = ElasticNet()
Elasticnet.fit(X_train_scaled,y_train)
y_pred = Elasticnet.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("R2 Score: ",score)
plt.scatter(y_test,y_pred)
plt.show()
```
